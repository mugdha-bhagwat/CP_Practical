```python
import math
c0 = 2.0 
k = 0.08 
t = 1

for i in range(1, 11):
    c = c0 - (k * c0 * t)
    print(f"Concentration of A at time {t} min: {c:.3f} mol/L")
    con_percent = ((c0- c) / c0 )* 100
    print(f"Percentage of A reacted at time {t} min: {con_percent:.2f}%")
    c0 = c
    t += 1

s = math.exp(-k * t) * c0
print(f"absolute error = {abs(s - c):.3f} mol/L")
```

```python
def fib_iter(n):
    series = []
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
        series.append(a)
    return series

print(fib_iter(10))  # Example usage: prints the first 10 Fibonacci numbers

def fib_rec(n):
    if n <= 1:
        return n
    else:
        return fib_rec(n-1) + fib_rec(n-2)  

print(fib_rec(10))  # Example usage: prints the 10th Fibonacci number


print("sample no. | fibonacci tag | sample time")
print([(f"{i} | {fib_iter(i)} | {fib_rec(i)}") for i in range(1, 11)])
```

```python
import pandas as pd

kinetics_df = pd.DataFrame({
    'Time_min': [0, 10, 20, 30, 40, 50],
    'run1_mol_l(low T)': [1.00, 0.74, 0.55, 0.41, 0.30, 0.22],
    'run2_mol_l(high T)': [1.00, 0.61, 0.37, 0.22, 0.14, 0.08]
})

kinetics_df
```

```python
#save it as kinetics.csv and kinetics.json

kinetics_df.to_csv('kinetics.csv')
kinetics_df.to_json('kinetics.json', orient='records')

read_csv = pd.read_csv('kinetics.csv')
print(read_csv.dtypes)
```

```python
run_columns = ['run1_mol_l(low T)', 'run2_mol_l(high T)']

print("Mean concentration by run:")
for column in run_columns:
    print(f"{column}: {kinetics_df[column].mean():.3f} mol/L")

print("\nSample standard deviation by run:")
for column in run_columns:
    print(f"{column}: {kinetics_df[column].std():.3f} mol/L")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(kinetics_df['Time_min'], kinetics_df['run1_mol_l(low T)'], marker='o', label='Run 1 (low T)')
plt.plot(kinetics_df['Time_min'], kinetics_df['run2_mol_l(high T)'], marker='s', label='Run 2 (high T)')
plt.title('Concentration vs Time')
plt.xlabel('Time (min)')
plt.ylabel('Concentration (mol/L)')
plt.legend()
plt.savefig('concentration_vs_time.png')
```

```python
import pandas as pd

comp_df = pd.DataFrame({
    'vol_l': [1,2,3,4,5],
    'pressure_bar': [4.00, 2.03, 1.31, 1.02, 0.81],
})

comp_df
```

```python
import matplotlib.pyplot as plt

volume = comp_df['vol_l']
pressure = comp_df['pressure_bar']
k = (pressure * volume).mean()
model_pressure = k / volume

plt.figure(figsize=(8, 5))
plt.scatter(volume, pressure, label='Measured pressure')
plt.plot(volume, model_pressure, marker='s', label=f'Boyle model: p = k/V (k = {k:.3f} bar L)')
plt.title('Pressure vs Volume')
plt.xlabel('Volume (L)')
plt.ylabel('Pressure (bar)')
plt.plot(comp_df['vol_l'], k/comp_df['vol_l'], linestyle='--', color='r', label='P = k/V')
plt.legend()
plt.grid(True)
plt.savefig('pressure_vs_volume.png')
```
