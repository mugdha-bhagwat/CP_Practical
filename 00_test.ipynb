```python
# if elif else statements for marks of students
# above 90 - A, 80-89 - C, 70-79 - D, 60-69 - E, below 60 - F
marks = float(input("Enter the marks: "))

if marks > 90:
    grade = "A"
elif marks >= 80:
    grade = "C"
elif marks >= 70:
    grade = "D"
elif marks >= 60:
    grade = "E"
else:
    grade = "F"

print("The grade is:", grade)
```
